```python
# 1. Import required modules
from anomalib.data import MVTecAD
from anomalib.engine import Engine
from anomalib.models import Patchcore
```

```python
!pip install -q -U kagglehub

import kagglehub
from pathlib import Path

data_dir = "/content/datasets/MVTecAD"

path = kagglehub.dataset_download(
    "ipythonx/mvtec-ad",
    path="bottle",
    output_dir=data_dir
)


print("Dataset baixado em:", path)

download_path = Path(path).resolve()
dataset_root = download_path.parent if download_path.name == "bottle" else download_path
print("Root usado pelo Anomalib:", dataset_root)
if not (dataset_root / "bottle").is_dir():
    raise FileNotFoundError(f"Categoria bottle não encontrada no caminho retornado: {download_path}")

```

```python

from lightning.pytorch.callbacks import TQDMProgressBar

datamodule = MVTecAD(
    root=dataset_root,
    category="bottle",
    train_batch_size=32,
    eval_batch_size=32,
    num_workers=2,
)

model = Patchcore()

engine = Engine(
    callbacks=[TQDMProgressBar()]
)

engine.fit(
    model=model,
    datamodule=datamodule
)
```

```python
# 6. Test the model performance
test_results = engine.test(datamodule=datamodule, model=model)
```
